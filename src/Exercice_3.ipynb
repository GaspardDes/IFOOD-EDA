¿Cómo varía el tipo de gasto en comida según los ingresos de los clientes?
Muestra visualmente que tipo de productos son los más consumidos por los
clientes según su nivel de ingresos.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/desp/Desktop/IFOOD-EDA/src/ml_project1_data.csv")
df.head()
```

Regroup in 3 categories (Low, Medium, High Incomes)

```python
df["Income_Group"] = pd.qcut(df["Income"], 3, labels=["Low", "Medium", "High"])
```

Showing means by categories for each group of incomes

```python
df.groupby("Income_Group")[["MntWines", "MntMeatProducts", "MntFishProducts", "MntGoldProds", "MntSweetProducts", "MntFruits"]].mean()
```

Graph showing spending by income group for each categories of spendings

```python
df.groupby("Income_Group")[["MntWines", "MntMeatProducts", "MntFishProducts", "MntGoldProds", "MntSweetProducts", "MntFruits"]].mean().plot(kind="bar")
plt.title("Spending by Income Group")
plt.show()
```

The global trend of spending for each categories is the same for all groups.
Perhaps, if we focus on particular categories like Wines, Meat and fish, we can see that significately increases with the level of incomes. That means that low income's group doesnt make their priorities buying Meat, Fish or Wine. Moreover we can see that Medium and High income's groups don't spend as much part of their money in GoldProds than Low income's group.

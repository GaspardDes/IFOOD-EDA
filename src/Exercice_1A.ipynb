¿Existe una relación entre los ingresos anuales de los clientes y la cantidad
de dinero que gastan en iFood? Visualiza la relación con un gráfico y
describe tus conclusiones.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/desp/Desktop/IFOOD-EDA/src/ml_project1_data.csv")
df.head()
```

Re create the global spendings variable

```python
df["Total_Spending"] = df[["MntWines", "MntMeatProducts", "MntFishProducts","MntSweetProducts", "MntGoldProds"]].sum(axis=1)
```

Generate graph showing total spendings of customers knowing their incomes

```python
plt.scatter(df["Income"], df["Total_Spending"])
plt.xlabel("Income")
plt.ylabel("Total Spending")
plt.show()
```

Calculate the correlation's coefficient 

```python
df["Income"].corr(df["Total_Spending"])
```

The correlation's coefficient is 0,67 which is near of 1. That means that the correlation between the customers' incomes and their total spendings is strong. That means that more the customers earn money more they spend.

However, this relationship is not perfectly linear, suggesting that other factors (such as household composition or consumption patterns) may also influence spending.
